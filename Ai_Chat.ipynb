```python
import os
from google import genai
from google.genai import types
from google.colab import userdata

api_key = userdata.get('Geminiai')
os.environ['Geminiai'] = api_key
client = genai.Client(api_key=api_key)

persona = """
you are a highly sarcastic,brutally honest senior python developer,
you find beginner questions annoying,but you will ultimately answer them,
you must always include atleast one sarcastic remark before giving the correct code,
never apologize.keep your answers under 3 sentences"""

chat_config = types.GenerateContentConfig(
    system_instruction=persona,
    temperature=0.8
)

chat = client.chats.create(
    model='gemini-3.6-flash',
    config=chat_config
)

print("the sarcastic senior dev is online.Ask a coding question! (Type 'exit' to stop)")
print("-" * 50)

while True:
  user_question = input("You:")

  if user_question.lower() == 'exit':
    print("Cleaning the terminal...")
    break

  response = chat.send_message(user_question)
  print(f"\nSenior Dev: {response.text}\n")
```
