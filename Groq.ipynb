```python
!pip install --upgrade groq

import os
from groq import Groq
from google.colab import userdata

# Retrieve your Groq API key
api_key = userdata.get('Groq')

# Initialize the official Groq client
client = Groq(api_key=api_key)

prompt = "Explain Quantum physics to a 5 year old in exactly two sentences."

print("Contacting Groq...\n")
response = client.chat.completions.create(
    model="llama-3.1-8b-instant",
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ]
)

print(response.choices[0].message.content)
```

```python
!pip install --upgrade google-genai

import os
from google import genai
from google.colab import userdata
from google.genai.errors import ServerError

api_key = userdata.get('Geminiai')
os.environ['Geminiai'] = api_key

client = genai.Client(api_key=api_key)

prompt = "Explain Quantum physics to a 5 year old in exactly two sentences."

print("Contacting the LLM...\n")

try:
    response = client.models.generate_content(
        model='gemini-3.8-flash',
        contents=prompt
    )
except ServerError as e:
    if e.code == 503:
        print("gemini-3.8-flash is currently unavailable due to high demand. Falling back to gemini-1.5-flash...\n")
        response = client.models.generate_content(
            model='gemini-1.5-flash',
            contents=prompt
        )
    else:
        raise e

print(response.text)
```
