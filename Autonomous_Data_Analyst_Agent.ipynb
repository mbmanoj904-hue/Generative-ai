```python

# ================================================================
# AUTONOMOUS DATA ANALYST AGENT
# WITHOUT GEMINI API / AI ANALYSIS
# GOOGLE COLAB - SINGLE CELL
# ================================================================

!pip -q install pandas numpy matplotlib seaborn scikit-learn openpyxl xlrd

import os
import json
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
from sklearn.ensemble import IsolationForest
from IPython.display import display, HTML

warnings.filterwarnings("ignore")

# ================================================================
# 1. UPLOAD DATASET
# ================================================================

print("=" * 70)
print("       AUTONOMOUS DATA ANALYST AGENT")
print("=" * 70)

print("\nUpload a CSV, Excel, JSON, HTML, or TXT dataset.")
uploaded = files.upload()

if not uploaded:
    raise ValueError("No file uploaded. Please upload a dataset.")

filename = next(iter(uploaded))
extension = os.path.splitext(filename)[1].lower()

# ================================================================
# 2. LOAD DATASET
# ================================================================

print("\nLoading dataset:", filename)

try:
    if extension == ".csv":
        df = pd.read_csv(filename)

    elif extension in [".xlsx", ".xls"]:
        df = pd.read_excel(filename)

    elif extension == ".json":
        df = pd.read_json(filename)

    elif extension in [".html", ".htm"]:
        tables = pd.read_html(filename)
        if not tables:
            raise ValueError("No tables found in the HTML file.")
        df = tables[0]

    elif extension == ".txt":
        try:
            df = pd.read_csv(filename, sep=None, engine="python")
        except Exception:
            df = pd.read_csv(filename, sep="\t")

    else:
        raise ValueError(
            "Unsupported file format. Use CSV, XLSX, XLS, JSON, HTML, or TXT."
        )

except Exception as error:
    raise ValueError(f"Could not load the dataset: {error}")

if df.empty:
    raise ValueError("The uploaded dataset is empty.")

print("\nDataset loaded successfully!")
print("Original rows:", len(df))
print("Original columns:", len(df.columns))

# ================================================================
# 3. DATA OVERVIEW
# ================================================================

print("\n" + "=" * 70)
print("1. DATASET OVERVIEW")
print("=" * 70)

print("\nFirst five rows:")
display(df.head())

print("\nDataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nOriginal missing values:")
print(df.isnull().sum())

# ================================================================
# 4. DATA CLEANING
# ================================================================

print("\n" + "=" * 70)
print("2. DATA CLEANING")
print("=" * 70)

df_clean = df.copy()

# Remove duplicate rows
duplicate_count = int(df_clean.duplicated().sum())
df_clean = df_clean.drop_duplicates()

# Remove completely empty columns
empty_columns = df_clean.columns[
    df_clean.isnull().all()
].tolist()

df_clean = df_clean.drop(columns=empty_columns)

# Clean column names
df_clean.columns = [
    str(column).strip() for column in df_clean.columns
]

# Convert numeric-looking text columns to numbers when possible
for column in df_clean.columns:
    if df_clean[column].dtype == "object":
        values = df_clean[column].astype(str).str.strip()
        converted = pd.to_numeric(
            values.str.replace(",", "", regex=False),
            errors="coerce"
        )

        if values.ne("").sum() > 0:
            valid_ratio = converted.notna().sum() / max(
                values.ne("").sum(), 1
            )

            if valid_ratio >= 0.8:
                df_clean[column] = converted

print("Duplicate rows removed:", duplicate_count)
print("Empty columns removed:", len(empty_columns))

# Fill missing values
for column in df_clean.columns:
    if df_clean[column].isnull().any():

        if pd.api.types.is_numeric_dtype(df_clean[column]):
            median_value = df_clean[column].median()

            if pd.notna(median_value):
                df_clean[column] = df_clean[column].fillna(
                    median_value
                )
            else:
                df_clean[column] = df_clean[column].fillna(0)

        else:
            mode_values = df_clean[column].mode(dropna=True)

            if not mode_values.empty:
                df_clean[column] = df_clean[column].fillna(
                    mode_values.iloc[0]
                )
            else:
                df_clean[column] = df_clean[column].fillna("Unknown")

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum())

print("\nCleaned dataset shape:", df_clean.shape)

# ================================================================
# 5. STATISTICAL ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("3. STATISTICAL ANALYSIS")
print("=" * 70)

numeric_columns = df_clean.select_dtypes(
    include=np.number
).columns.tolist()

if numeric_columns:
    statistics = df_clean[numeric_columns].describe().T
    display(statistics)
else:
    statistics = pd.DataFrame()
    print("No numeric columns found.")

# ================================================================
# 6. CORRELATION ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("4. CORRELATION ANALYSIS")
print("=" * 70)

correlations = []

if len(numeric_columns) >= 2:
    correlation_matrix = df_clean[numeric_columns].corr()

    display(correlation_matrix)

    for i, col1 in enumerate(numeric_columns):
        for col2 in numeric_columns[i + 1:]:
            value = correlation_matrix.loc[col1, col2]

            if pd.notna(value):
                correlations.append({
                    "Feature_1": col1,
                    "Feature_2": col2,
                    "Correlation": round(float(value), 3)
                })

    correlations.sort(
        key=lambda item: abs(item["Correlation"]),
        reverse=True
    )

    print("\nStrongest correlations:")
    for item in correlations[:10]:
        print(
            f"{item['Feature_1']} vs {item['Feature_2']}: "
            f"{item['Correlation']}"
        )
else:
    correlation_matrix = pd.DataFrame()
    print("At least two numeric columns are required.")

# ================================================================
# 7. ANOMALY DETECTION
# ================================================================

print("\n" + "=" * 70)
print("5. ANOMALY DETECTION")
print("=" * 70)

df_clean["Anomaly"] = "Normal"
outlier_count = 0

if len(numeric_columns) >= 1 and len(df_clean) >= 10:

    anomaly_data = df_clean[numeric_columns].replace(
        [np.inf, -np.inf], np.nan
    )

    # Use median values for missing or infinite numeric values
    anomaly_data = anomaly_data.fillna(
        anomaly_data.median()
    ).fillna(0)

    if len(anomaly_data) >= 10:
        detector = IsolationForest(
            contamination=0.05,
            random_state=42
        )

        predictions = detector.fit_predict(anomaly_data)

        df_clean["Anomaly"] = np.where(
            predictions == -1,
            "Anomaly",
            "Normal"
        )

        outlier_count = int((predictions == -1).sum())

        print("Potential anomalies detected:", outlier_count)
        display(df_clean[df_clean["Anomaly"] == "Anomaly"].head(20))

else:
    print(
        "Anomaly detection requires at least one numeric column "
        "and 10 rows."
    )

# ================================================================
# 8. AUTOMATIC BUSINESS METRIC DETECTION
# ================================================================

print("\n" + "=" * 70)
print("6. BUSINESS METRICS")
print("=" * 70)

business_metrics = {}
column_lookup = {
    str(column).lower().strip(): column
    for column in df_clean.columns
}

metric_keywords = {
    "Revenue": ["revenue", "sales", "total_sales", "amount", "turnover"],
    "Profit": ["profit", "net_profit", "earnings"],
    "Quantity": ["quantity", "qty", "units_sold", "units"],
    "Price": ["price", "unit_price", "selling_price"],
    "Rating": ["rating", "review_score", "satisfaction_score"]
}

for metric, keywords in metric_keywords.items():
    for column in df_clean.columns:
        normalized = str(column).lower().strip()

        if any(keyword in normalized for keyword in keywords):
            if pd.api.types.is_numeric_dtype(df_clean[column]):
                business_metrics[metric] = float(
                    df_clean[column].sum()
                    if metric in ["Revenue", "Profit", "Quantity"]
                    else df_clean[column].mean()
                )
                print(
                    f"{metric}: {business_metrics[metric]:,.2f}"
                )
                break

if not business_metrics:
    print("No common business metrics were automatically identified.")

# ================================================================
# 9. CATEGORY ANALYSIS
# ================================================================

print("\n" + "=" * 70)
print("7. CATEGORY ANALYSIS")
print("=" * 70)

categorical_columns = df_clean.select_dtypes(
    include=["object", "category"]
).columns.tolist()

# Exclude the generated anomaly label
categorical_columns = [
    col for col in categorical_columns if col != "Anomaly"
]

category_summaries = {}

for column in categorical_columns[:5]:
    print(f"\nTop values in {column}:")
    counts = df_clean[column].value_counts().head(10)
    display(counts)

    category_summaries[column] = counts.to_dict()

# ================================================================
# 10. VISUALIZATION
# ================================================================

print("\n" + "=" * 70)
print("8. DATA VISUALIZATION")
print("=" * 70)

if numeric_columns:
    for column in numeric_columns[:5]:
        plt.figure(figsize=(8, 4))
        sns.histplot(
            df_clean[column].dropna(),
            kde=True
        )
        plt.title(f"Distribution of {column}")
        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.tight_layout()
        plt.show()

if len(numeric_columns) >= 2:
    plt.figure(figsize=(10, 6))
    sns.heatmap(
        correlation_matrix,
        annot=True,
        cmap="coolwarm",
        fmt=".2f"
    )
    plt.title("Correlation Heatmap")
    plt.tight_layout()
    plt.show()

for column in categorical_columns[:2]:
    plt.figure(figsize=(9, 4))
    df_clean[column].value_counts().head(10).plot(
        kind="bar"
    )
    plt.title(f"Top Categories: {column}")
    plt.xlabel(column)
    plt.ylabel("Count")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

# ================================================================
# 11. RULE-BASED BUSINESS RECOMMENDATIONS
# ================================================================

print("\n" + "=" * 70)
print("9. BUSINESS RECOMMENDATIONS")
print("=" * 70)

recommendations = []

if duplicate_count > 0:
    recommendations.append(
        f"Review data collection processes: "
        f"{duplicate_count} duplicate rows were removed."
    )

missing_before = int(df.isnull().sum().sum())

if missing_before > 0:
    recommendations.append(
        f"Investigate missing data: {missing_before} missing "
        f"values existed in the original dataset."
    )

if outlier_count > 0:
    recommendations.append(
        f"Review the {outlier_count} potential anomalies before "
        f"making business decisions."
    )

if "Revenue" in business_metrics:
    recommendations.append(
        f"Total detected revenue-like metric: "
        f"{business_metrics['Revenue']:,.2f}. "
        f"Compare results across products, regions, or time periods "
        f"when those columns are available."
    )

if "Profit" in business_metrics:
    recommendations.append(
        f"Total detected profit-like metric: "
        f"{business_metrics['Profit']:,.2f}. "
        f"Investigate products or categories with lower profitability."
    )

if "Rating" in business_metrics:
    recommendations.append(
        f"Average detected rating: "
        f"{business_metrics['Rating']:.2f}. "
        f"Review low-rated products or customer feedback if available."
    )

if correlations:
    strongest = correlations[0]
    recommendations.append(
        f"The strongest numeric correlation is between "
        f"{strongest['Feature_1']} and {strongest['Feature_2']} "
        f"({strongest['Correlation']}). Investigate this relationship; "
        f"correlation does not prove causation."
    )

if not recommendations:
    recommendations.append(
        "Explore the dataset further and define business-specific "
        "metrics to generate more targeted recommendations."
    )

for index, recommendation in enumerate(recommendations, 1):
    print(f"{index}. {recommendation}")

# ================================================================
# 12. GENERATE TEXT REPORT - NO EXTERNAL AI
# ================================================================

print("\n" + "=" * 70)
print("10. BUSINESS ANALYSIS REPORT")
print("=" * 70)

report_lines = [
    "AUTONOMOUS DATA ANALYST REPORT",
    "=" * 60,
    f"Dataset: {filename}",
    f"Original rows: {len(df)}",
    f"Original columns: {len(df.columns)}",
    f"Cleaned rows: {len(df_clean)}",
    f"Cleaned columns: {len(df_clean.columns)}",
    f"Duplicate rows removed: {duplicate_count}",
    f"Empty columns removed: {len(empty_columns)}",
    f"Potential anomalies detected: {outlier_count}",
    "",
    "BUSINESS METRICS",
    "-" * 40
]

if business_metrics:
    for key, value in business_metrics.items():
        report_lines.append(f"{key}: {value:,.2f}")
else:
    report_lines.append("No recognized business metrics found.")

report_lines.extend([
    "",
    "BUSINESS RECOMMENDATIONS",
    "-" * 40
])

for index, recommendation in enumerate(recommendations, 1):
    report_lines.append(f"{index}. {recommendation}")

report_lines.extend([
    "",
    "STRONGEST CORRELATIONS",
    "-" * 40
])

if correlations:
    for item in correlations[:10]:
        report_lines.append(
            f"{item['Feature_1']} vs {item['Feature_2']}: "
            f"{item['Correlation']}"
        )
else:
    report_lines.append("No correlations available.")

ai_report = "\n".join(report_lines)

print(ai_report)
print("\nReport generated using data analysis rules. No AI API was used.")

# ================================================================
# 13. CREATE HTML DASHBOARD
# ================================================================

print("\nCreating HTML dashboard...")

overview_html = df_clean.head(20).to_html(
    index=False,
    border=0,
    classes="data-table"
)

metrics_html = pd.DataFrame(
    list(business_metrics.items()),
    columns=["Metric", "Value"]
).to_html(
    index=False,
    border=0,
    classes="data-table"
)

recommendations_html = "".join(
    f"<li>{recommendation}</li>"
    for recommendation in recommendations
)

html_content = f"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<title>Autonomous Data Analyst Dashboard</title>
<style>
body {{
    font-family: Arial, sans-serif;
    margin: 30px;
    background: #f4f6f9;
    color: #222;
}}
h1, h2 {{ color: #17365d; }}
.card {{
    background: white;
    padding: 20px;
    margin: 15px 0;
    border-radius: 10px;
    box-shadow: 0 2px 6px #ddd;
    overflow-x: auto;
}}
.data-table {{
    border-collapse: collapse;
    width: 100%;
}}
.data-table th, .data-table td {{
    border: 1px solid #ddd;
    padding: 8px;
    text-align: left;
}}
.data-table th {{ background: #e5edf7; }}
</style>
</head>
<body>
<h1>Autonomous Data Analyst Dashboard</h1>

<div class="card">
<h2>Dataset Overview</h2>
<p><b>File:</b> {filename}</p>
<p><b>Original rows:</b> {len(df)}</p>
<p><b>Cleaned rows:</b> {len(df_clean)}</p>
<p><b>Cleaned columns:</b> {len(df_clean.columns)}</p>
<p><b>Duplicates removed:</b> {duplicate_count}</p>
<p><b>Potential anomalies:</b> {outlier_count}</p>
</div>

<div class="card">
<h2>Business Metrics</h2>
{metrics_html}
</div>

<div class="card">
<h2>Recommendations</h2>
<ol>{recommendations_html}</ol>
</div>

<div class="card">
<h2>Cleaned Dataset Preview</h2>
{overview_html}
</div>

<div class="card">
<h2>Analysis Report</h2>
<pre>{ai_report}</pre>
</div>

<p>Generated locally using Python data-analysis libraries.</p>
</body>
</html>
"""

with open("data_analyst_dashboard.html", "w", encoding="utf-8") as file:
    file.write(html_content)

print("HTML dashboard created.")

# ================================================================
# 14. SAVE CLEANED DATASET
# ================================================================

print("\nSaving cleaned datasets...")

df_clean.to_csv(
    "cleaned_dataset.csv",
    index=False
)

df_clean.to_excel(
    "cleaned_dataset.xlsx",
    index=False
)

df_clean.to_json(
    "cleaned_dataset.json",
    orient="records",
    indent=2,
    force_ascii=False
)

print("Created:")
print("- cleaned_dataset.csv")
print("- cleaned_dataset.xlsx")
print("- cleaned_dataset.json")

# ================================================================
# 15. SAVE TEXT REPORT
# ================================================================

with open(
    "data_analysis_report.txt",
    "w",
    encoding="utf-8"
) as file:
    file.write(ai_report)

print("\nText report created: data_analysis_report.txt")

# ================================================================
# 16. FINAL SUMMARY
# ================================================================

print("\n" + "=" * 70)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("Dataset:", filename)
print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))
print("Numeric columns:", len(numeric_columns))
print("Categorical columns:", len(categorical_columns))
print("Anomalies detected:", outlier_count)
print("Recommendations generated:", len(recommendations))

print("\nAvailable outputs:")
print("1. cleaned_dataset.csv")
print("2. cleaned_dataset.xlsx")
print("3. cleaned_dataset.json")
print("4. data_analysis_report.txt")
print("5. data_analyst_dashboard.html")

# ================================================================
# 17. DOWNLOAD OUTPUT FILES
# ================================================================

print("\nDownloading output files...")

for output_file in [
    "cleaned_dataset.csv",
    "cleaned_dataset.xlsx",
    "cleaned_dataset.json",
    "data_analysis_report.txt",
    "data_analyst_dashboard.html"
]:
    files.download(output_file)

print("\nAll tasks completed!")
```
